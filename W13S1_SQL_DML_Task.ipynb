# Setup

```python
# 1. Install pandasql
!pip install pandasql --quiet

# 2. Load libraries and your CSV into a pandas DataFrame
import pandas as pd
from pandasql import sqldf

url = "https://docs.google.com/spreadsheets/d/e/2PACX-1vRa1wWwXmzxEvqITxj4OQTeLywlGTTsOTbhSRqKj2lPuGefjlci-DQhgLBPpgWXe8AAUu2WUBqY59X1/pub?gid=1030172542&single=true&output=csv"
df = pd.read_csv(url)   # ← replace with your file
df.head()

# 3. Create a helper to run SQL against any DataFrame in your notebook
pysqldf = lambda query: sqldf(query, globals())
```

# 1. Show records of 'male' patient from 'southwest' region.

```python
pysqldf("""
SELECT * FROM df
WHERE gender = 'male' and region = 'southwest'
""")
```

# 2. Show all records having bmi in range 30 to 45 both inclusive.

```python
pysqldf("""
SELECT * FROM df
WHERE bmi BETWEEN 30 and 45
""")
```

# 3. Find no of unique patients who are not from southwest region.

```python
print(pysqldf("""
SELECT COUNT(DISTINCT PatientID) FROM df
WHERE region != "southwest"

"""))
```

```python
df['PatientID'].nunique()
```

# 4. Show minimum and maximum bloodpressure of diabetic patient who smokes. Make column names as MinBP and MaxBP respectively.

```python
pysqldf("""
SELECT MAX(bloodpressure) AS 'MaxBP',Min(bloodpressure) AS 'MinBP' FROM df
WHERE diabetic = 'Yes' and smoker = 'Yes'
""")
```

# 5.Total claim amount from male smoker.

```python
pysqldf("""
SELECT SUM(claim)AS 'claim_amount' FROM df
WHERE gender = 'male'
""")
```

# 6.Select all records of south region.

```python
pysqldf("""
SELECT * FROM df
WHERE region LIKE 'south%'
""")
```

# 7.No of patient having normal blood pressure. Normal range[90-120]

```python
pysqldf("""
SELECT COUNT(PatientID) FROM df
WHERE bloodpressure BETWEEN 90 AND 120
""")
```

# 8. No of pateint belo 17 years of age having normal blood pressure as per below formula -

BP normal range = 80+(age in years × 2) to 100 + (age in years × 2)

```python
pysqldf("""
SELECT COUNT(PatientID) FROM df
WHERE bloodpressure BETWEEN 80+age*2 AND 100+age*2

""")
```

# 9. What is the average claim amount for non-smoking female patients who are diabetic?

```python
pysqldf("""
SELECT AVG(claim) FROM df
WHERE gender = 'female' and smoker = 'No'
""")
```

# 10.Write a SQL query to update the claim amount for the patient with PatientID = 1234 to 5000.

```python
pysqldf("""
UPDATE df
SET claim = 5000
WHERE PatienID = 1234
""")
```

# 11.Write a SQL query to delete all records for patients who are smokers and have no children.
