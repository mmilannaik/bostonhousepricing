### **`Problem-2:`** Write a Python function that accepts a hyphen-separated sequence of words as parameter and returns the words in a hyphen-separated sequence after sorting them alphabetically.

**Example 1:**

Input:
```bash
green-red-yellow-black-white
```

Output:
```bash
black-green-red-white-yellow
```

````python
text = "green-red-yellow-black-white"
print(sorted(text.split('-')))
````

````python
# Write code here
text = "green-red-yellow-black-white"
def seq_sep(text):
  temp = sorted(text.split("-"))
  res = ''
  counter = 0
  for i in temp:
    if counter == 0:
      res = res+i
      counter+=1
    else:
      res = res+"-"+i
      counter+=1
  return res
````

````python
seq_sep(text)
````

### **`Problem 3:`** Write a Python function that accepts a string and calculate the number of upper case letters and lower case letters.

```
Sample String : 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'
Expected Output :
No. of Upper case characters :  9
No. of Lower case Characters :  47
```

````python
# Write code here

def calc_case(text):
  upper_count,lower_count = 0,0
  for word in text:
    for i in word:
      #print(i)
      if i.isupper():
        upper_count +=1
      elif i.islower():
        lower_count +=1
      else:
        pass
  return f"uppercount is {upper_count},lower count is {lower_count}"
````

````python
text = 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'
calc_case(text)
````

### **`Problem 5:`** Write a Python function to check whether a number is perfect or not.

A Perfect number is a number that is half the sum of all of its positive divisors (including itself).

Example :

```
The first perfect number is 6, because 1, 2, and 3 are its proper positive divisors, and 1 + 2 + 3 = 6.
Equivalently, the number 6 is equal to half the sum of all its positive divisors: ( 1 + 2 + 3 + 6 ) / 2 = 6.

The next perfect number is 28 = 1 + 2 + 4 + 7 + 14. This is followed by the perfect numbers 496 and 8128.
```


````python
# Write code here

def check_per(num):
  temp = []
  sum = 0
  for i in range(1,num):
    if num%i==0:
      temp.append(int(i))
  print(temp)
  for i in range(0,len(temp)):
    sum = sum +temp[i]
  #total = sum(temp)
  print(sum)
  if sum == num and (sum+num)//2 == num :
    return 'It is a perfect number'
  else:
    return 'It is not a perfect number'
````

### **`Problem-6:`** Write a Python function to concatenate any no of dictionaries to create a new one.

```
Sample Dictionary :
dic1={1:10, 2:20}
dic2={3:30, 4:40}
dic3={5:50,6:60}
Expected Result : {1: 10, 2: 20, 3: 30, 4: 40, 5: 50, 6: 60}
```

````python
# Write code here
def dict_concat(*args):
  res = {}
  for key,val in args.items():
    res.append(key,val)
  return res
````

````python
dic1={1:10, 2:20}
dic2={3:30, 4:40}
dic3={5:50,6:60}
print(type(dic1))
dict_concat(dic1,dic2,dic3)
````

`Problem-8` Write a python function that receives a list of integers and prints out a histogram of bin size 10

```
Input:
[13,42,15,37,22,39,41,50]
```

```
Output:
{11-20:2,21-30:1,31-40:2,41-50:3}
```

````python
# Write code here
def binsize(num):
  hist_size = ['11-20','21-30','31-40','41-50']
  c1,c2,c3,c4 = 0,0,0,0
  for i in num:
    if i>=11 and i<=20:
      c1 += 1
    elif i>=21 and i<=30:
      c2 += 1
    elif i>=31 and i<=40:
      c3 += 1
    else:
      c4 += 1
  hist_num = []
  hist_num.extend([c1,c2,c3,c4])
  return dict(zip(hist_size,hist_num))

````

````python
num = [13,42,15,37,22,39,41,50]
binsize(num)
````
