# NGram Model

Language modeling is the way of determining the probability of any sequence of words. Language modeling is used in various applications such as Speech Recognition, Spam filtering, etc. Language modeling is the key aim behind implementing many state-of-the-art Natural Language Processing models.

Methods of Language Modelling
Two methods of Language Modeling:

Statistical Language Modelling: Statistical Language Modeling, or Language Modeling, is the development of probabilistic models that can predict the next word in the sequence given the words that precede. Examples such as N-gram language modeling.
Neural Language Modeling: Neural network methods are achieving better results than classical methods both on standalone language models and when models are incorporated into larger models on challenging tasks like speech recognition and machine translation. A way of performing a neural language model is through word embeddings.
N-gram
N-gram can be defined as the contiguous sequence of n items from a given sample of text or speech. The items can be letters, words, or base pairs according to the application. The N-grams typically are collected from a text or speech corpus (A long text dataset).

For instance, N-grams can be unigrams like (“This”, “article”, “is”, “on”, “NLP”) or bigrams (“This article”, “article is”, “is on”, “on NLP”).

N-gram Language Model
An N-gram language model predicts the probability of a given N-gram within any sequence of words in a language. A well-crafted N-gram model can effectively predict the next word in a sentence, which is essentially determining the value of p(w∣h), where h is the history or context and w is the word to predict.

# Code- Ngram Model

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

```python

# Sample dataset: short text snippets in different languages
data = {
    'text': [
        'hello', 'thanks', 'bye',         # English
        'hola', 'gracias', 'adiós',       # Spanish
        'bonjour', 'merci', 'salut',      # French
        'hallo', 'danke', 'tschüss',      # German
        'ciao', 'grazie', 'addio'         # Italian
    ],
    'language': [
        'en', 'en', 'en',
        'es', 'es', 'es',
        'fr', 'fr', 'fr',
        'de', 'de', 'de',
        'it', 'it', 'it'
    ]
}

df = pd.DataFrame(data)

# Split into train and test
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['language'], test_size=0.3, random_state=42)

# Character n-gram model using scikit-learn pipeline
char_ngram_model = Pipeline([
    ('vectorizer', CountVectorizer(analyzer='char', ngram_range=(2, 4))),
    ('classifier', MultinomialNB())
])

char_ngram_model.fit(X_train, y_train)
y_pred = char_ngram_model.predict(X_test)

# Display evaluation
report = classification_report(y_test, y_pred, output_dict=True)
```

```python
report
```

```python
report_df = pd.DataFrame(report).transpose()
```

```python
from gensim.models.phrases import Phrases,Phraser
from gensim.models import FastText
import pandas as pd
```

 the parameters min_count and threshold in Phrases() directly influence which word combinations become multi-word expressions (like chronic_kidney).

Let’s break it down:

🔧 Phrases(sentences, min_count=1, threshold=1): What Do They Mean?
Parameter	Description
min_count	Minimum number of times a word pair must occur together to be considered a phrase
threshold	Controls how strong the association between the words must be. Lower = more phrases

🔍 Example
Let’s say your sentence list is:


sentences = [
    ["chronic", "kidney", "disease"],
    ["chronic", "kidney", "infection"],
    ["chronic", "kidney", "failure"],
    ["chronic", "pain"]
]

'🔹 Case 1: Phrases(sentences, min_count=2, threshold=10)
"chronic kidney" appears 3 times ✅ (min_count=2)

But if threshold=10, it may be too strict, so chronic_kidney may not be created.

🔹 Case 2: min_count=1, threshold=1
Even rare co-occurrences (1 time!) are accepted

"chronic_kidney", "kidney_failure", "chronic_pain" → all become phrases

🎯 Tuning Tips for Medical Texts
Scenario	Recommended Setting
Very small corpus	min_count=1, threshold=1–5
Medium dataset	min_count=3–5, threshold=5–15
Very large dataset (millions)	min_count=5–10, threshold=15–30

✅ Summary: Practical Effects
Setting	Effect
min_count=1, threshold=1	Very aggressive, finds lots of phrases (even noisy ones)
min_count=5, threshold=10	Balanced — keeps only meaningful, repeated phrases
min_count=10, threshold=20	Conservative — only common & strong phrases kept

```python
sent = [row.split() for row in df['Text']]
phrases = Phrases(sent,min_count = 30)
sentences = phrases[sent]

```

```python
# intitializing the model
model = FastText(sentences = sentences,
                 vector_size = 100,
                 window = 5,
                 min_count = 5,
                 workers = 4,
                 min_n = 1,
                 max_n = 4)
```

```python
# Building vocab
model.build_vocab(sentences)
```

```python
# Training the model
model.train(sentences,total_examples = len(sentences),epochs=100)
```

```python
# saving the model
import joblib
path = 'Fasttext.joblib'
joblib.dump(model,path)
```

```python
model.wv.most_similar('python',topn=5)
```

```python
model.wv.most_similar("epidemic out-break", topn = 10)
```

```python
import langcodes

def get_language_name(label):
    code = label.replace("__label__", "")
    return langcodes.get(code).display_name()

# Example
label, prob = model_lang.predict("Bonjour tout le monde")
print(get_language_name(label[0]))  # Output: French
```

```python
texts = ["Hello", "नमस्ते", "Bonjour", "Hallo", "Ciao"]

for t in texts:
    label, prob = model_lang.predict(t)
    lang = get_language_name(label[0])
    print(f"Text: {t}\n→ Detected: {lang} ({prob[0]:.2f} confidence)\n")
```
